### Create Voive Trigger

```python
# Clear history
chat_session = model.start_chat(
#  history=[]
)
```

```python
import os
import speech_recognition as sr
import pyttsx3
from pocketsphinx import LiveSpeech
import time

# Initialize text-to-speech engine
engine = pyttsx3.init()

# Function to speak out text
def speak(text):
    engine.say(text)
    engine.runAndWait()

# Function to recognize speech from microphone using PocketSphinx
def recognize_speech():
    recognizer = sr.Recognizer()
    mic = sr.Microphone()

    # Speak the prompt and wait for it to finish
    prompt_text = "How can I help you? I am listening to you...."
    print(prompt_text)
    speak(prompt_text)
    
    # Adding a small delay to ensure prompt is fully played before listening
    time.sleep(0.5)
    
    with mic as source:
        recognizer.adjust_for_ambient_noise(source)
        print("Listening for your response...")
        audio = recognizer.listen(source)

    try:
        print("Recognizing speech...")

        # ========= Choose the STT method for use: (The google online SR doesn't	work on Sam's Mac)
        # speech_text = recognizer.recognize_google(audio) # Use Google Speech Recognition to recognize speech (online)
        # ========= Choose the STT method for use:
        speech_text = recognizer.recognize_sphinx(audio) # Use PocketSphinx to recognize speech (offline)

        print("\nYou said: \n", speech_text)
        return speech_text
    except sr.UnknownValueError:
        print("Sorry, I could not understand the audio.")
        return ""
    except sr.RequestError as e:
        print("Could not request results from PocketSphinx; {0}".format(e))
        return ""

# Function to simulate API call
def call_gemini_api(query):
    # Simulating API call to Gemini
    print(f"\nSending recognized speech to Gemini API: {query}")
    r = chat_session.send_message(query)  # Make sure the chat_session has been defined (in the code above)
    return r.text

def main():
    # Recognize user's speech
    user_input = recognize_speech()

    if user_input:
        # Send the recognized speech to the Gemini API
        response_text = call_gemini_api(user_input)

        # Display and speak the response from the API
        print("\nAPI Response: \n", response_text)
        speak(response_text)
    else:
        print("No speech recognized.")

if __name__ == "__main__":
    main()
```
