```python
api_key = input("api key : ")
```

```python
import google.generativeai as palm

palm.configure(api_key=api_key)

model = { 'model': "models/chat-bison-001"}

q = input("please enter your text : ")
```

```python
r = palm.chat(
  **model,
  messages=q
)

print(r.last)
```
