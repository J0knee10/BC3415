```python
from transformers import ViTFeatureExtractor, ViTForImageClassification
from PIL import Image
import requests
```

```python
#url = 'http://images.cocodataset.org/val2017/000000039769.jpg'
#image = Image.open(requests.get(url, stream=True).raw)

image = Image.open("/content/drive/MyDrive/AI4AnF/Data/Taylor_Swift_(6966830273).jpg")
```

```python
image
```

```python
feature = ViTFeatureExtractor.from_pretrained('google/vit-base-patch16-224')
model = ViTForImageClassification.from_pretrained('google/vit-base-patch16-224')

inputs = feature(images=image, return_tensors="pt")
pred = model(**inputs)
pred = pred.logits
# model predicts one of the 1000 ImageNet classes
pred = pred.argmax(-1).item()
print("Predicted class:", model.config.id2label[pred])
```
