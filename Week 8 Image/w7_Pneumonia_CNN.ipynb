```python
train_dir = "/content/drive/MyDrive/TT Library/AI Model/Data/Pneumonia/train"
```

```python
test_dir = "/content/drive/MyDrive/TT Library/AI Model/Data/Pneumonia/test"
```

```python
from keras.preprocessing.image import ImageDataGenerator
```

```python
train_data = ImageDataGenerator().flow_from_directory(train_dir, (150,150), batch_size=5, shuffle=False)
```

```python
test_data = ImageDataGenerator().flow_from_directory(test_dir, (150,150), batch_size=5, shuffle=False)
```

```python
from keras.models import Sequential
```

```python
from keras.layers import Conv2D,MaxPool2D,Dropout,Flatten,Dense
```

```python
model = Sequential()
```

```python
model.add(Conv2D(32,(3,3),input_shape=(150,150,3),activation="relu"))
```

```python
model.add(MaxPool2D(pool_size=(2,2),strides=2))
model.add(Dropout(0.1))
```

```python
model.add(Conv2D(32,(3,3),activation="relu"))
model.add(MaxPool2D(pool_size=(2,2),strides=2))
model.add(Dropout(0.1))
```

```python
model.add(Conv2D(64,(3,3),activation="relu"))
model.add(MaxPool2D(pool_size=(2,2),strides=2))
model.add(Dropout(0.1))
```

```python
model.add(Conv2D(64,(3,3),activation="relu"))
model.add(MaxPool2D(pool_size=(2,2),strides=2))
model.add(Dropout(0.1))
```

```python
model.add(Flatten())
```

```python
model.add(Dense(128,activation="relu"))
model.add(Dropout(0.1))
```

```python
model.add(Dense(2,activation="softmax"))
```

```python
model.summary()
```

```python
model.compile(loss="categorical_crossentropy",optimizer="adam",metrics=["accuracy"])
```

```python
h = model.fit(train_data, batch_size=5, epochs=5)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(h.history["loss"])
```

```python
model.evaluate(test_data)
```
